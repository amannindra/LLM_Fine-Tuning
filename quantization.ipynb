## Int8 Quantization

https://www.youtube.com/watch?v=IIvORO248Zs

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1.0, 1.0, 100)
x[:10]
```

```python
min_x = -1
max_x = 1
qmin = -128
qmax = 127

scale = (max_x - min_x) / (qmax - qmin)
scale
```

```python
zero_point = 0
q = np.round((x - zero_point) / scale).astype(int)
print(q)
q = np.clip(q, qmin, qmax)
print(q)
```

## Reverse Quantization

```python
x_hat = scale * (q - zero_point)
x_hat
```

```python




average = abs(x - x_hat)
print(average)

add = np.sum(average) / len(average)
add
```

## NF4


```python
nf4_values = np.array([-1.0, -0.75, -0.5, -0.35, -0.2, -0.1, -0.05, -0.01, 0.01, 0.05, 0.1, 0.2, 0.35, 0.5, 0.75, 1])


x = np.linspace(-1.0, 1.0, 100)
x[:10]
```

```python
def quantize_to_nf4(x_vals, nf4_lut):
    indices = np.argmin(np.abs(x_vals[:, None] - nf4_lut[None, :]), axis=1)
    return nf4_lut[indices]

x_nf4 = quantize_to_nf4(x, nf4_values)
x_nf4[:10]
```
